```python
# Swap first and last element of the list

def swap_first_last(lst):
    if len(lst) > 1:
        lst[0], lst[-1] = lst[-1], lst[0]
    return lst

numbers = [1, 2, 3, 4, 5]
print(swap_first_last(numbers)) 
```

```python
#Count length of string

def count_length(s):
    return len(s)

string = "Hello, World!"
print(count_length(string))  
```

```python
#  Sum of non-negative integers

def sum_non_negative(numbers):
    return sum(num for num in numbers if num >= 0)

numbers = [1, 4, -5, -20, 10]
print(sum_non_negative(numbers)) 
```

```python
# write a code of Factorial

def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers")
    elif n == 0:
        return 1
    else:
        return n * factorial(n-1)

number = 3
print(factorial(number))  
```
