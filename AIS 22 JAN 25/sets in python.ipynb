```python
# Create a set
numbers = {1, 2, 3, 4, 5}
```

```python
# Add an element
numbers.add(6)
print("Set after add:", numbers)
```

```python

# Remove an element
numbers.remove(4)
print("Set after remove:", numbers)
```

```python
# Union
numbers_union = numbers.union({6, 7, 8})
print("Union set:", numbers_union)
```

```python
# Intersection
numbers_intersection = numbers.intersection({4, 5, 6})
print("Intersection set:", numbers_intersection)
```

```python
# Difference
numbers_difference = numbers.difference({5, 6})
print("Difference set:", numbers_difference)
```
