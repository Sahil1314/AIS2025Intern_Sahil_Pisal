```python
# Indexing and Slicing

# # Accessing specific elements or subsets of an array.


import numpy as np

# Create a 2D array
array = np.array([[1, 2, 3], [4, 5, 6]])

# Access the first element
print(array[0, 0])
```

```python
# Access the last element
print(array[-1, -1]) 
```

```python
# Access a subset of elements (slicing)
print(array[0, 1:]) 
```

```python
# Masking

# Selecting elements based on conditions using logical indexing.


import numpy as np

# Create an array
array = np.array([1, 2, 3, 4, 5])

# Select elements greater than 3
mask = array > 3
print(array[mask]) 
```

```python
# Fancy Indexing

# Using arrays of indices to select elements.


import numpy as np

# Create an array
array = np.array([1, 2, 3, 4, 5])

# Select elements at indices 1 and 3
indices = np.array([1, 3])
print(array[indices]) 
```

```python
# These examples demonstrate how to perform indexing and slicing, masking, and fancy indexing on NumPy arrays.

# Advanced Indexing

# Using multiple arrays of indices to select elements.


import numpy as np

# Create a 2D array
array = np.array([[1, 2, 3], [4, 5, 6]])

# Select elements at indices (0, 1) and (1, 2)
row_indices = np.array([0, 1])
col_indices = np.array([1, 2])
print(array[row_indices, col_indices])
```
