```python
# Integers
a = 10
b = 20
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
# Floats
c = 10.5
d = 20.7
print("Addition:", c + d)
print("Subtraction:", c - d)
print("Multiplication:", c * d)
print("Division:", c / d)
```

```python
# Complex numbers
e = 10 + 5j
f = 20 + 7j
print("Addition:", e + f)
print("Subtraction:", e - f)
print("Multiplication:", e * f)
print("Division:", e / f)
```
