```python
# Universal Functions (ufuncs)
# Element-wise operations like np.add(), np.subtract(), np.multiply(), np.divide().


import numpy as np

# Create two arrays
array1 = np.array([1, 2, 3, 4, 5])
array2 = np.array([6, 7, 8, 9, 10])

# Addition
result = np.add(array1, array2)
print(result)  
```

```python
# Subtraction
result = np.subtract(array1, array2)
print(result) 
```

```python
# Multiplication
result = np.multiply(array1, array2)
print(result) 
```

```python
# Division
result = np.divide(array1, array2)
print(result)  
```

```python
# Statistical Functions
# Calculating mean, median, standard deviation, variance, minimum, and maximum values.


import numpy as np

# Create an array
array = np.array([1, 2, 3, 4, 5])

# Mean
result = np.mean(array)
print(result) 
```

```python
# Median
result = np.median(array)
print(result) 
```

```python
# Standard Deviation
result = np.std(array)
print(result)  
```

```python
# Variance
result = np.var(array)
print(result)  
```

```python
# Minimum
result = np.min(array)
print(result) 
```

```python
# Maximum
result = np.max(array)
print(result) 
```

```python
# Linear Algebra Functions
# Performing operations like matrix multiplication, inversion, and eigenvalue decomposition.


import numpy as np

# Create two matrices
matrix1 = np.array([[1, 2], [3, 4]])
matrix2 = np.array([[5, 6], [7, 8]])

# Matrix Multiplication
result = np.dot(matrix1, matrix2)
print(result)
```

```python
# Matrix Inversion
result = np.linalg.inv(matrix1)
print(result)
```

```python
# Eigenvalue Decomposition
eigenvalues, eigenvectors = np.linalg.eig(matrix1)
print(eigenvalues)
# Output: [-0.37228132  5.37228132]
print(eigenvectors)
```
