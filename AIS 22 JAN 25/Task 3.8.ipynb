```python
def count_case_letters(word):
    lowercase_count = sum(1 for char in word if char.islower())
    uppercase_count = sum(1 for char in word if char.isupper())
    return lowercase_count, uppercase_count

word = 'STatiStiCS'
lowercase_count, uppercase_count = count_case_letters(word)
print(f'Lowercase count: {lowercase_count}')
print(f'Uppercase count: {uppercase_count}')
```
