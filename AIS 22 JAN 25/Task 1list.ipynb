```python
# Create a list
fruits = ["apple", "banana", "cherry"]

# Accessing elements
print("First fruit:", fruits[0])
print("Last fruit:", fruits[-1])
print("slicing:",fruits[1:2])   #starts from 1 and end to index 2 does not returns end i.e index 2 element
# Modifying elements
fruits[0] = "mango"
print("Modified list:", fruits)
```

```python
# Adding elements
fruits.append("orange")
print("List after append:", fruits)
```

```python
fruits.insert(2, "grape")
print("List after insert:", fruits)
```

```python
# Removing elements
fruits.remove("banana")
print("List after remove:", fruits)
```

```python
fruits.pop(0)
print("List after pop:", fruits)
```

```python
# Other methods
fruits.sort()
print("Sorted list:", fruits)
```

```python
fruits.reverse()
print("Reversed list:", fruits)
```

```python
fruits.extend(["pear", "peach"])
print("List after extend:", fruits)
```

```python
print("Count of 'apple':", fruits.count("apple"))
```

```python
print("Index of 'cherry':", fruits.index("cherry"))
```

```python
fruits_copy = fruits.copy()
print("Copied list:", fruits_copy)
```
