```python
# Create a dictionary
person = {"name": "sahil", "age": 21, "city": "New York"}

# Accessing key-value pairs
print("Name:", person["name"])
print("Age:", person["age"])
```

```python
# Adding a key-value pair
person["country"] = "INDIA"
print("Dictionary after add:", person)
```

```python
# Removing a key-value pair
del person["city"]
print("Dictionary after remove:", person)
```

```python
# Get method
print("Name:", person.get("name"))
```

```python
# Keys method
print("Keys:", person.keys())
```

```python

# Values method
print("Values:", person.values())
```

```python
# Items method
print("Items:", person.items())
```

```python
# Pop method
print("Popped item:", person.pop("age"))
```
