```python
import numpy as np

# Create two arrays
array1 = np.array([1, 2, 3, 4, 5])
array2 = np.array([6, 7, 8, 9, 10])

# Addition
result = array1 + array2
print(result)  
```

```python
# substraction
result = array1 - array2
print(result)  
```

```python
# Multiplication
result = array1 * array2
print(result)  
```

```python
# Division
result = array1 / array2
print(result)
```

```python
# BROADCASTING
import numpy as np

# Create two arrays with different shapes
array1 = np.array([1, 2, 3])
array2 = np.array([4])

# Addition
result = array1 + array2
print(result) 
```

```python
# Multiplication
result = array1 * array2
print(result)
```

```python
# ELEMENT WISE OPERATIONS AND FUNCTIONS
import numpy as np

# Create an array
array = np.array([1, 4, 9, 16, 25])

# Square root
result = np.sqrt(array)
print(result)
```

```python
# Sine
result = np.sin(array)
print(result) 
```

```python
# Exponential
result = np.exp(array)
print(result)  
```
