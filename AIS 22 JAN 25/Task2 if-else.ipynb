```python
# Task 2.1: Grade Code

def grade_code(p):
    if p > 75:
        return "O"
    elif 60 <= p <= 75:
        return "A"
    elif 50 <= p < 60:
        return "B"
    elif 35 <= p < 50:
        return "C"
    else:
        return "Fail"
print(grade_code(50))
```

```python
def divisibility_check(n):
    if n % 2 == 0 and n % 3 == 0:
        return f"{n} is divisible by both 2 and 3"
    elif n % 2 == 0 and n % 3 != 0:
        return f"{n} is divisible by 2, but not by 3"
    elif n % 2 != 0 and n % 3 == 0:
        return f"{n} is divisible by 3, but not by 2"
    else:
        return f"{n} is not divisible by either 2 or 3"
print(divisibility_check(13))
```

```python
# if elif else examples
# # Example 1: age check
a = int(input("Enter your age: "))

# If elif else ladder
if(a>=18):
    print("You are above the age of consent")
    print("Good for you")

elif(a<0):
    print("You are entering an invalid negative age")

elif(a==0):
    print("You are entering 0 which is not a valid age")    

else:
    print("You are below the age of consent")


print("End of Program")
```

```python
# Example 2: Grade Calculator

def calculate_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

print(calculate_grade(85)) 
```

```python
# Example 3: Day of the Week

def find_day(day_number):
    if day_number == 1:
        return "Monday"
    elif day_number == 2:
        return "Tuesday"
    elif day_number == 3:
        return "Wednesday"
    elif day_number == 4:
        return "Thursday"
    elif day_number == 5:
        return "Friday"
    elif day_number == 6:
        return "Saturday"
    elif day_number == 7:
        return "Sunday"
    else:
        return "Invalid day number"

print(find_day(3)) 
```

```python
# Example 4: Age Validator

def validate_age(age):
    if age < 18:
        return "You are a minor."
    elif age < 65:
        return "You are an adult."
    else:
        return "You are a senior citizen."

print(validate_age(25))  
```

```python
# Example 5: Temperature Checker

def check_temperature(temp):
    if temp < 0:
        return "Freezing"
    elif temp < 20:
        return "Cold"
    elif temp < 30:
        return "Warm"
    else:
        return "Hot"

print(check_temperature(25))  
```
