```python
# Arithmetic operators
a = 10
b = 20
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("reminder:",a%b)
```

```python
# Comparison operators
print("Equal:", a == b)
print("Not Equal:", a != b)
print("Less than:", a < b)
print("Greater than:", a > b)
```

```python
# Logical operators
print("And:", a > 5 and b < 25) #returns true if both are true
print("Or:", a > 5 or b < 25)  #true if one of them is true
print("not:",not( a > 5 or b < 25))  #return opposite result
```

```python
# Assignment operators
a += 5                     # Increment the value of b by 3 and then assign it to a
print("Addition assignment:", a)
```
