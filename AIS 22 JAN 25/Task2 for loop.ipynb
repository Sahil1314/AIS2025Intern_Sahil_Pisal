```python
# for loop
# Print odd values between 20 and 80 without using if

for i in range(20, 80):
    if i % 2 != 0: 
        print(i)
```

```python
 # Create a list of numbers from 1 to 20 using for loop

numbers = []
for i in range(1, 21):
    numbers.append(i)
print(numbers)
```

```python
#  Create a list of numbers from 20 to 1 using for loop

numbers = []
for i in range(20, 0, -1):
    numbers.append(i)
print(numbers)

```

```python
#  Take cube of odd values between 20 to 40

cubes = []
for i in range(20, 40):
    if i % 2 != 0:
        cubes.append(i ** 3)
print(cubes)

```

```python
# Task 5: Print friend names with corresponding ages

names = ["a", "b", "c", "d", "e"]
ages = [20, 21, 23, 25, 24]

for i in range(len(names)):
    print(f"My name is {names[i]}, my age is {ages[i]}")
```
