```python
# Odd-Even using While Loop


num = 1
while num <= 10:
    if num % 2 == 0:
        print(num, "is even")
    else:
        print(num, "is odd")
    num += 1
```

```python
i = 0
while i < 5:
    print("* " * i)
    i += 1
```

```python
# Task 3: Create list 1-20 numbers using While Loop
numbers = []
i = 1
while i <= 20:
    numbers.append(i)
    i += 1
print(numbers)
```

```python
# Create list 20-1 (reverse order) using While Loop


numbers = []
i = 20
while i >= 1:
    numbers.append(i)
    i -= 1
print(numbers)
```

```python
#  Using Break Control Statement


i = 0
while i < 10:
    if i == 5:
        break
    print(i)
    i += 1
print("Loop exited")
```
