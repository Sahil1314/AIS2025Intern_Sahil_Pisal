```python
# Create a tuple
colors = ("red", "green", "blue")
```

```python
# Accessing elements
print("First color:", colors[0])
print("Last color:", colors[-1])
```

```python
# Slicing
print("Middle colors:", colors[1:2])
```

```python
# Concatenation
colors += ("yellow", "purple")
print("Concatenated tuple:", colors)
```

```python
# Repetition
colors_repeated = colors * 2
print("Repeated tuple:", colors_repeated)
```

```python
# Count
print("Count of 'green':", colors.count("green"))
```

```python
# Index
print("Index of 'blue':", colors.index("blue"))
```
